#### 2.4 Setting Up The Validation Framework

###### Machine Learning Zoomcamp — Setting Up the Validation Framework

## Recap: Why We Split the Data

As covered in the previous session, to validate a model properly, the dataset needs to be split into **three parts**:

1. **Train** — used to train the model
2. **Validation** — used to check how well the model performs while developing it
3. **Test** — used only occasionally, at the very end, to confirm the model actually generalizes

From each of these three splits, we derive:

- A **feature matrix** `X`
- A **target variable** `y`

This gives us: `X_train`/`y_train`, `X_val`/`y_val`, and `X_test`/`y_test`.

## Step 1: Decide the Split Sizes

The plan is a **60 / 20 / 20** split (train / validation / test).

```python
n = len(df)  # ~12,000 records

n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test  # avoids rounding mismatches
```

**Why not just compute `n_train = int(n * 0.6)` directly?** Rounding each piece independently can cause the three sizes to not add back up to `n` (some records could get silently dropped). Computing `n_train` as *"whatever's left over"* after subtracting validation and test guarantees the numbers always sum correctly.

Resulting split (for \~12,000 records):

- Train: \~7,000
- Validation: \~2,400
- Test: \~2,400

## Step 2: Slicing the DataFrame with `.iloc`

`.iloc` supports range-style slicing, similar to plain Python lists:

```python
df.iloc[:10]        # first 10 records (indices 0–9)
df.iloc[10:20]       # records 10 through 19
df.iloc[10:]         # everything from index 10 to the end
```

A naive first approach: take the first `n_val` rows for validation, the next `n_test` rows for test, and the rest for train.

**Problem:** the dataset has an inherent order (e.g., records might be grouped by manufacturer). Slicing sequentially like this could mean validation ends up with all the BMWs and none appear in train — an unrepresentative split.

## Step 3: Shuffle Before Splitting

To avoid any accidental ordering in the data, shuffle the row indices before slicing.

```python
import numpy as np

idx = np.arange(n)      # sequence of indices: 0 to n-1
np.random.shuffle(idx)  # shuffle in place
```

Now use `.iloc[idx[...]]` to pull rows in shuffled order:

```python
df_train = df.iloc[idx[:n_train]]
df_val   = df.iloc[idx[n_train:n_train+n_val]]
df_test  = df.iloc[idx[n_train+n_val:]]
```

## Step 4: Make It Reproducible with a Random Seed

Without a fixed seed, you'll get a different shuffle — and therefore different train/val/test splits — every time you run the code, which makes results hard to compare or reproduce.

```python
np.random.seed(2)   # any fixed number works; 2 is just an example

idx = np.arange(n)
np.random.shuffle(idx)
```

With the same NumPy version and the same seed, the shuffle (and therefore the split) will be identical every time you run it — on your machine or anyone else's.

## Step 5: Reset the Index

After shuffling, each split's DataFrame retains its *original* row indices in a jumbled order, which is inconvenient.

```python
df_train = df_train.reset_index(drop=True)
df_val   = df_val.reset_index(drop=True)
df_test  = df_test.reset_index(drop=True)
```

`drop=True` prevents the old index from being kept as a new column — it's simply discarded since it's no longer needed.

## Step 6: Prepare the Target Variable (`y`)

The target column (`MSRP`, in this project) needs a **log transformation** (`log1p`) applied, and should be extracted as a plain NumPy array rather than kept as a Pandas Series (since the index isn't needed anymore):

```python
y_train = np.log1p(df_train.msrp.values)
y_val   = np.log1p(df_val.msrp.values)
y_test  = np.log1p(df_test.msrp.values)
```

## Step 7: Remove the Target Column from the Feature Data

Once `y` has been extracted, the target column must be **deleted** from each DataFrame:

```python
del df_train['msrp']
del df_val['msrp']
del df_test['msrp']
```

**Why this matters:** if the target column (`msrp`) accidentally stays in the feature matrix, the model can "cheat" by using price to predict price — producing a suspiciously perfect result. This is a common and easy-to-miss bug, so it's good practice to explicitly delete the target column right after separating it out.

## Summary

In this lesson, the validation framework was implemented **manually**, using only Pandas and NumPy (no library shortcuts):

1. Compute split sizes (60/20/20), avoiding rounding errors
2. Shuffle row indices with a fixed random seed for reproducibility
3. Slice the shuffled DataFrame into train/validation/test
4. Reset each split's index
5. Extract and log-transform the target variable (`y`)
6. Delete the target column from the feature DataFrames to avoid data leakage

#### 2.5 Linear Regression

Linear regression is a model used for **regression problems** — predicting a number as output (as opposed to classification, which predicts a category, or ranking, which orders items).

Recall the general formula from the introduction:

```
g(X) ≈ y
```

- `g` — the model
- `X` — the feature matrix
- `y` — the target (in this project, car price)

Linear regression is one specific choice for `g`.

## Simplifying: Looking at One Observation

Instead of the entire feature matrix `X`, it helps to first look at a **single row** — one car — treated as a vector:

```
x_i = [x_i1, x_i2, ..., x_in]
```

This is an `n`-dimensional vector representing one car's features. The goal is a function that takes this vector and produces a number close to that car's actual price.

## A Concrete Example

Using the **training set** (never validation or test data during training), take row 10: a *Rolls-Royce Phantom Drophead Coupe (2015)*.

To keep things simple, only three features are used:

| Feature | Value |
| --- | --- |
| Engine horsepower | 453 |
| City miles per gallon | 11 |
| Popularity | 86 |

So for this car:

```python
x_i = [453, 11, 86]
```

The goal is a function `g(x_i)` that returns a prediction — e.g., "this car costs $10,000" — which then needs to actually be implemented.

## The Linear Regression Formula

```
g(x_i) = w0 + w1·x_i1 + w2·x_i2 + w3·x_i3
```

- **`w0`** — the **bias term**: the prediction made with *no* information about the car at all (a baseline/average).
- **`w1, w2, w3`** — weights, one per feature, controlling how much each feature pushes the prediction up or down.

This can be written more compactly as a sum:

```
g(x_i) = w0 + Σ(wⱼ · x_ij)   for j = 1 to n
```

(In code, since indexing starts at 0, the sum runs from `j = 0` to `n - 1`.)

## Implementing It in Code

```python
def linear_regression(xi):
    n = len(xi)
    
    pred = w0
    for j in range(n):
        pred = pred + w[j] * xi[j]
    
    return pred
```

Where:

- `w0` is the bias term
- `w` is a vector of weights, one per feature (same length as `xi`)

## Running the Example

Using example weights:

- `w0 = 7.17`
- `w1 = 0.01` (horsepower)
- `w2 = 0.04` (city mpg)
- `w3 = 0.002` (popularity)

With `x_i = [453, 11, 86]`, the prediction comes out to **12.3**.

### Interpreting the Result

- **`w0 = 7.17`** — the baseline prediction for an "average" car with no other information.
- **Horsepower (`w1 = 0.01`)** — more horsepower increases predicted price. Makes intuitive sense: more powerful engines tend to cost more.
- **City MPG (`w2 = 0.04`)** — counterintuitively, *higher* fuel consumption (lower efficiency) is associated with a *higher* price here — likely because fancier, more powerful cars tend to consume more fuel.
- **Popularity (`w3 = 0.002`)** — a very small weight, meaning popularity (e.g., social media mentions) has only a minor effect on price unless the number of mentions is very large (the instructor speculates a car like a Tesla might see a bigger effect from this feature).

## Undoing the Log Transformation

Recall that the target `y` was log-transformed earlier (`log1p`) before training. The value `12.3` above is therefore **not** the actual dollar prediction — it's in log space.

To convert back to a real price, apply the inverse of `log1p`, which is `expm1`:

```python
import numpy as np

price = np.expm1(12.3)
```

This gives a predicted price of approximately **$222,000** for this car.

> `np.expm1` and `np.log1p` are inverses of each other — applying one after the other returns the original value.

## Summary

- Linear regression combines a bias term (`w0`) with a weighted sum of the input features to produce a prediction.
- The formula was implemented manually — one feature vector at a time — without using a library.
- Predictions need to be converted back from log space (using `expm1`) to get an actual interpretable price.
- **Next up:** generalizing this from a single example to the *entire* dataset at once (working with the full matrix `X` rather than one row `x_i` at a time).
