```python
original_median = df['horsepower'].median()
print(original_median) 
```

```python
mode_value = df['horsepower'].mode()[0] 
print(mode_value)
```

```python
df['horsepower_filled'] = df['horsepower'].fillna(mode_value) 
new_median = df['horsepower_filled'].median() 
print(new_median)
```

```python
print(f"Original: {original_median}, After filling: {new_median}")

if new_median > original_median:
    print("Yes, it increased")
elif new_median < original_median:
    print("Yes, it decreased")
else:
    print("No")
```

```python
df_asia = df[df['origin'] == 'Asia'] # Select all cars from Asia
df_asia
```

```python
df_asia = df_asia[['vehicle_weight', 'model_year']]  # Only vehicle weight & model_year columns
df_asia
```

```python
df_asia = df_asia.head(7) # First Seven values
df_asia
```

```python
X = df_asia.values # NumPy array (X)
X
```

```python
XTX = X.T.dot(X) #XTX = Xᵀ · X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX) # Inverse of the matrix
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200]) # y array
y
```

```python
w = XTX_inv.dot(X.T).dot(y) # Compute w = (XᵀX)⁻¹ · Xᵀ · y
```

```python
float(w.sum())
```
